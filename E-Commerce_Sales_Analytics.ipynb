```python
missing = df.isnull().sum()

missing_percentage = (missing / len(df)) * 100

missing_data = pd.DataFrame({
    'Missing Values': missing,
    'Percentage': missing_percentage
})

missing_data[missing_data['Missing Values'] > 0]
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
num_cols = df.select_dtypes(include=np.number).columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
num_cols
```

```python
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    if df[col].notna().any():
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
cat_cols
```

```python
df['customer_age'] = df['customer_age'].fillna(
    df['customer_age'].median()
)
```

```python
df['customer_age']
```

```python
df['customer_review'] = df['customer_review'].fillna('No Review')

df['review_sentiment'] = df['review_sentiment'].fillna('Unknown')
```

```python
df['review_sentiment']
```

```python
df['customer_review']
```

```python
df.loc[
    df['return_status'].isna() & df['return_reason'].isna(),
    'return_reason'
] = 'Not Applicable'
```

```python
df['order_date'] = pd.to_datetime(
    df['order_date'], errors='coerce'
)
```

```python
df['order_date']
```

```python
df['Year'] = df['order_date'].dt.year

df['Month'] = df['order_date'].dt.month_name()

df['Month_Number'] = df['order_date'].dt.month
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(x=df['net_sales'])

plt.title('Net Sales Outlier Analysis')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(x=df['profit'])

plt.title('Profit Outlier Analysis')

plt.show()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])

df['Month'] = df['order_date'].dt.month_name()

monthly_sales = df.groupby('Month')['net_sales'].sum()

plt.figure(figsize=(12,5))

plt.plot(monthly_sales.index, monthly_sales.values, marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Revenue')

plt.xticks(rotation=45)

plt.show()
```
